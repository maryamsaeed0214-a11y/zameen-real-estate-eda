# Feature Engineering

```python
df["Price"].dtype
```

```python
df["Bedrooms"].dtype
```

```python
df["Bedrooms"] = pd.to_numeric(df["Bedrooms"], errors="coerce")
```

```python
df["Price"].head()
```

```python
df["Price"].dtype
```

```python
df["Price"] = df["Price"].str.replace("PKR", "", regex=False)
df["Price"] = df["Price"].str.replace("\n", " ", regex=False)
df["Price"] = df["Price"].str.strip()
```

```python
def convert_price(price):
    if pd.isna(price):
        return None

    price = str(price).strip()

    if "Crore" in price:
        return float(price.replace("Crore", "").strip()) * 10000000

    elif "Lakh" in price:
        return float(price.replace("Lakh", "").strip()) * 100000

    elif "Thousand" in price:
        return float(price.replace("Thousand", "").strip()) * 1000

    else:
        try:
            return float(price)
        except:
            return None
```

```python
df["Price"] = df["Price"].apply(convert_price)
```

```python
df["Price"].head()
```

```python
df["Price"].dtype
```

```python
df["Bedrooms"] = pd.to_numeric(df["Bedrooms"], errors="coerce")
```

```python
df["Price_per_Bedroom"] = df["Price"] / df["Bedrooms"]
```

# Data Visualization

```python
plt.figure(figsize=(10,5))
sns.histplot(df["Price"], bins=30, kde=True)
plt.title("Distribution of Property Prices")
plt.xlabel("Price (PKR)")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.histplot(df["Area"], bins=30, kde=True)
plt.title("Distribution of Property Area")
plt.xlabel("Area (Sq. Yd.)")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.countplot(y=df["City"], order=df["City"].value_counts().index)
plt.title("Number of Properties by City")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.countplot(y=df["Type"], order=df["Type"].value_counts().index)
plt.title("Property Types")
plt.show()
```

```python
plt.figure(figsize=(10,6))
df["Location"].value_counts().head(10).plot(kind="bar")
plt.title("Top 10 Property Locations")
plt.xlabel("Location")
plt.ylabel("Number of Properties")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10,6))
df.groupby("City")["Price"].mean().sort_values().plot(kind="bar")
plt.title("Average Property Price by City")
plt.ylabel("Average Price (PKR)")
plt.xticks(rotation=45)
plt.show()
```

```python
numeric_df = df.select_dtypes(include=["int64", "float64"])
```

```python
plt.figure(figsize=(10,8))
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(x="Bedrooms", y="Price", data=df)
plt.title("Price vs Bedrooms")
plt.show()
```
